```python
import numpy as np

special_tokens = [
    "[PAD]", 
    "[UNK]", 
    "<BOP>", # begining of problem
    "<EOP>", # end of problem
    "<BOQ>", # begining of questions
    "<EOQ>", # end of questions
]

full_problems = np.concat(df["Full Problem"].apply(set).apply(list).values)
questions = np.concat(df["Possible Questions"].apply(set).apply(list).values)

vocabulary = set(np.concat([full_problems, questions, special_tokens]))
vocab_size = len(vocabulary)
```

```python
data = list(df["Full Problem"].values) + list(df["Possible Questions"].values)
data[:5]
```

```python
from tokenizers import Tokenizer, AddedToken
from tokenizers.models import WordLevel
from tokenizers.trainers import WordLevelTrainer
from tokenizers import pre_tokenizers

tokenizer = Tokenizer(WordLevel(unk_token="[UNK]"))

tokenizer.add_special_tokens([
    AddedToken(t, single_word=False, lstrip=False, rstrip=False, normalized=False)
    for t in ["<BOP>", "<EOP>", "<BOQ>", "<EOQ>"]
])

tokenizer.pre_tokenizer = pre_tokenizers.Split("", behavior="isolated")

trainer = WordLevelTrainer(
    min_frequency=1,
    special_tokens=special_tokens
)


tokenizer.train_from_iterator(data, trainer)

tokenizer.enable_padding(
    pad_token="[PAD]",
    pad_id=tokenizer.token_to_id("[PAD]")
)
```

```python
enc = tokenizer.encode(
    df["Rewritten Problem"].apply(
        lambda x: f"<BOP>{x}".replace("What is", "<EOP><BOQ>").replace("?", "<EOQ>"))[0]
)
print(enc.tokens)
```
